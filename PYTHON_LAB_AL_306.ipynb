```python
# SCIPY
import numpy as np
from scipy import optimize, stats, interpolate


# Curve fitting example
# define a function to fit
def model_func(x, a, b, c):
  return a * np.sin(b*x) + c

# Generate some synthetic data
np.random.seed(0)
x_data = np.linspace(0,10,100)
y_data = model_func(x_data, 2.5, 1.3, 0.5) + np.random.normal(size=x_data.size)

# Fit the model to the data
params, params_covariance =optimize.curve_fit(model_func, x_data, y_data, p0=[2,1,0.5])
print("fitted parameter", params)
# Statistical functions example
# Calculate mean and standard deviation
data = np.random.normal(loc=5, scale=2, size=100)
mean = np.mean(data)
std_dev = np.std(data)
print("mean:", mean)
print("standard deviation:", std_dev)

# Interpolation example
# Generate some synthetic data for interpolation
x_interp = np.linspace(0, 10, 10)
y_interp = np.cos(x_interp)

# Perform linear interpolation
interp_func = interpolate.interp1d(x_interp, y_interp)
x_new = np.linspace(0, 10, 10)
y_new = interp_func(x_new)

# Statistical test example
# Perform a t-test
sample1 = np.random.normal(loc=5, scale=2, size=100)
sample2 = np.random.normal(loc=4, scale=2, size=100)
t_statistic, p_value = stats.ttest_ind(sample1, sample2)
print("p-value:", p_value)
```

```python
def day_name(day):
  match day:
    case 1:
      return "Monday"
    case 2:
      return "Tuesday"
    case 3:
      return "Wednesday"
    case 4:
      return "Thursday"
    case 5:
      return "Friday"
    case 6:
      return "Saturday"
    case 7:
      return "Sunday"
    case _:
      return "Invalid day"
print(day_name(5))
print(day_name(8))
```

```python
def fruit_color(fruits):
  match fruits:
    case "apple" | "cherry":
      return "Red"
    case "banana" | "lemon":
      return "Yellow"
    case "grape":
      return "Purple"
    case _:
      return "Unknown color"

print(fruit_color("apple"))
print(fruit_color("banana"))
print(fruit_color("grape"))
print(fruit_color("orange"))
```

```python
def process_point(point):
  match point:
    case (0,0):
      return "origin"
    case (x,0):
      return f"X-axis at {x}"
    case (0,y):
      return f"Y-axis at {y}"
    case (x,y):
      return f"X={x}, Y={y}"
    case _:
      return "Not a point"
print(process_point((0,0)))
print(process_point((3,0)))
print(process_point((0,5)))
print(process_point((2,4)))
print(process_point("abc"))
```

**TO WRITE A PROGRAM TO FIND THE SQUARE ROOTOF A NUMBER BY NEWTON'S METHOD**.

```python
def newtonsqrt(n):
  approx = 0.5 * n
  better = 0.5 * (approx + n/approx)
  while better != approx:
    approx = better
    better = 0.5 * (approx + n/approx)
  return approx
n = int(input("enter a number:"))
print("the square root of", n, "is approximately", newtonsqrt(n))
```

**Linear Search:-**
Linear search (or sequential search) is the simplest search algorithm. It checks each element one by one.

Implement Linear search in python, the fastest way check if a value exists in a list is to use in the operator.


```python
# Check if a value exist in a list:
mylist = [3,7,2,9,5,1,8,4,6]

if 4 in mylist:
  print("Found")
else:
  print("Not Found!")
```

```python
# Find the index of a value in a list:
def LinearSearch(arr,targetVal):
  for i in range(len(arr)):
    if arr[i] == targetVal:
      return i
  return -1

mylist = [3,7,2,9,5,1,8,4,6]
targetVal = 4
result = LinearSearch(mylist,targetVal)
if result != -1:
  print("Found at index", result)
else:
  print("Not Found!")
```

Searches for an element in a list anmd return its position.

```python
def linear_search(arr, target):
  for i in range(len(arr)):
    if arr[i] == target:
        return i     # Return index if found
  return -1        # Return -1 if not found

# Example
arr = [10,23,45,70,11,15]
target = 70

result = linear_search(arr, target)

if result != -1:
  print("Element found at index", result)
else:
  print("Element not found in the list")
```

```python
# Linear Search Using for element in array
def linear_search(arr, target):
  for index, value in enumerate(arr):
    if value == target:
      return index
  return -1

# Example
print(linear_search([3, 8, 2, 7, 5], 7))
```

```python
# Linear Search Returning Boolen (True/False)
def linear_search(arr, target):
  return target in arr

# Example
print(linear_search([4, 9, 1, 2], 1))
print(linear_search([4, 9, 1, 2], 7))
```

```python
# Linear Search in a String
def search_char(text, char):
    for i in range(len(text)):
        if text[i] == char:
            return i
    return -1
print(search_char("python", "h"))
```

```python
# Linear Search For All Occurance
def linear_search_all(arr, target):
    result = []
    for i in range(len(arr)):
         if arr[i] == target:
              result.append(i)
    return result

# Example
print(linear_search_all([5, 2, 5, 7, 5], 5))
```

```python
# SELECTION SORT
def selectionSort(alist):
    for i in range(len(alist)-1,0,-1):
        pos = 0
        for location in range(1,i+1):
            if alist[location]>alist[pos]:
                pos = location
        temp = alist[i]
        alist[i] = alist[pos]
        alist[pos] = temp
alist = [54,26,93,17,77,31,44,55,20]
selectionSort(alist)
print(alist)
```

**MERGE SORT**

```python
def merge(arr, l, m, r):
    n1 = m - l + 1
    n2 = r - m

    # create temp arrays
    L = [0] * (n1)
    R = [0] * (n2)

    # Copy data to temp arrays L[] and R[]
    for i in range(0, n1):
        L[i] = arr[l + i]
    for j in range(0, n2):
        R[j] = arr[m + 1 + j]

    # Merge the temp arrays back into arr[l..r]
    i = 0     # Initial index of first subarray
    j = 0     # Initial index of second subarray
    k = l     # Initial index of merged subarray

    while i < n1 and j < n2:
        if L[i] <= R[j]:
            arr[k] = L[i]
            i += 1
        else:
            arr[k] = R[j]
            j += 1
        k += 1

    # Copy the remaining elements of L[], if there are any
    while i < n1:
        arr[k] = L[i]
        i += 1
        k += 1

    # Copy the remaining elements of R[], if there are any
    while j < n2:
        arr[k] = R[j]
        j += 1
        k += 1
```
