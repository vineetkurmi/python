```python
# 1.) Single Inheritance
class Parent:
    def _init_(self):
        print('This is Parent class constructor')

    def greet(self):
        print('This is Parent class')

class Child(Parent):
    def _init_(self):
        print("This is child class constructor")

    def show(self):
        print('This is Child class')


obj = Child()
obj.greet()
obj.show()
```

```python
class Factory:
    def _init_(self,name,color):
        self.name = name
        self.color = color

    def show(self):
        print(f'Bag has {self.name} and {self.color} color')

class Bata(Factory):
    def _init_(self,name,color,zip,pockets):
        super()._init_(name,color)
        self.zip = zip
        self.pockets = pockets

    def display(self):
        print(f'Bag has {self.name} , {self.color} color , {self.zip} zip and {self.pockets} pockets')
Rahul = Bata('Rahul','Purple',4,10)
Rahul.display()
```

```python
#2. Multiple Inheritane -> 2 Parent, 1 Child

class Father: #Parent1

    def _init_(self):
        print('This is Father class constructor')

    def greet_father(self):
        print('This is Father class')

class Mother: #Parent2
    def _init_(self):
        print('This is Mother class constructor')

    def greet_mother(self):
        print('This is Mother class')


class Child(Mother,Father): #Child
    #If we have to run constructor of Father class first

    def _init_(self):
        Father._init_(self) #Sabse pehle Father class ka constructor will be run
        Mother._init_(self) #After Father class Mother class constructor will be run

obj = Child()
obj.greet_father()
obj.greet_mother()
```

```python
# 3.) Multilevel Inheritance. -> one child becomes the parent of another class

class A:  # Parent class
  def greet(self):
    print("This is class A.")

class B(A):  # Child class
  def show(self):
    print("This is class B.")

class C(B):  # Child class
  def details(self):
    print("This is class C.")

obj = C()
obj.greet()
obj.show()
obj.details()
```

```python
class CEO:     # Parent Class
  def __init__(self):
    print('This is CEO class constructor')

class Manager(CEO):    # Child Class , Parent class
  def __init__(self):
    super().__init__()
    print('This is Manager class constructor')

class Employee(Manager):   # Child Class
  def __init__(self):
    super().__init__()
    print('This is Employee class constructor')

rahul = Employee()
```

```python
class Account:
  def __init__(self,name,balance):
    self.name = name
    self.balance = balance

  def details(self):
    print(f'Hello {self.name} you have {self.balance}')

class Saving(Account):
  def __init__(self,name,balance):
    super().__init__(name,balance)
    print(f'This is saving class constructor {self.name} , {self.balance}')

class Current(Account):
  def __init__(self,name,balance,type):
    super().__init__(name,balance)
    self.type = type
    print(f'This is current class constructor {self.name} , {self.balance} , {self.type}')

obj = Current("Mukesh" , 0 , "Current")
```

```python
# Create sutdent admission management

class Students:
  def __init__(self,name,age,mail,number):
    self.name = name
    self.age = age
    self.mail = mail
    self.number = number
  def display_details(self):
    print(self.name)
    print(self.age)
    print(self.mail)
    print(self.number)


class Class10Admission(Students):
  def __init__(self,name,age,mail,number):
    super().__init__(name,age,mail,number)

    print("Admission successful in class 10")

class Class12Admission(Students):
  def __init__(self,name,age,mail,number):
    super().__init__(name,age,mail,number)

    if self.age >= 16:
      print("Admission successful in class 12")
    else:
      print("Admission Failed!")

print("Press 1 for Class 10 Admission")
print("Press 2 for Class 12 Admission")

choice = int(input("Enter your choice: "))

name = input("Enter your name: ")
age = int(input("Enter your age: "))
mail = input("Enter your mail: ")
number = input("Enter your number: ")

if choice == 1:
  students1 = Class10Admission(name,age,mail,number)
  students1.display_details()

if choice == 2:
  students1 = Class12Admission(name,age,mail,number)
  students1.display_details()
```

**Overridding**

```python
class Reebok:
  def __init__(self,material,size):
    self.material = material
    self.size = size

  def details(self):
    print("Details of bags")
    print(self.material)
    print(self.size)

class Campus(Reebok):
  def __init__(self,material,size,color):
    super().__init__(material,size)
    self.color = color

  def details(self):
    print(super().details())
    print(self.color)

obj1 = Campus("Leather",10,"Black")
obj1.details()

# A child class object has the power to call methods & attributes of parent
# class but it cannot call the details method of his parent class cause that
# details methods is overriddden and this concept is known as overridding.
```

**Abstraction**

```python
from abc import ABC, abstractmethod

class Shapes(ABC):
  @abstractmethod
  def area(self):
    pass

  @abstractmethod
  def perimeter(self):
    pass

class Square(Shapes):
  def __init__(self,side):
    self.side = side

  def area(self):
    print(f'Area of square is {self.side * self.side}')

  def perimeter(self):
    print(f'Perimeter of square is {4 * self.side}')

class Circle(Shapes):
  def __init__(self,radius):
    self.radius = radius

  def area(self):
    print(f'Area of circle is{3.14 * self.radius * self.radius}')

  def perimeter(self):
    print(f'Perimeter of circle is {2 * 3.14 * self.radius}')

obj = Square(10)
obj.area()
obj.perimeter()

obj1 = Circle(10)
obj1.area()
obj1.perimeter()
```

```python
class Numbers:
  def __init__(self,value)
    self.value = value

  def __add__(self,other):
    return self.value + other.value

  def __sub__(self,other):
    return self.value - other.value

  def __mul__(self,other):
    return self.value * other.value

  def __eq__(self,value):
    return self.value == value.value

a = Numbers(10)
b = Numbers(20)
print(a + b)
print(a - b)
print(a * b)
print(a == b)
```
