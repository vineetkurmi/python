**Breadth First Search (BFS)**

```
graph={
    '5' : ['3','7'],
    '3' : ['2', '4'],
    '7' : ['8'],
    '2' : [],
    '4' : ['8'],
    '8' : []
}

def bfs(visited, graph, node):
  visited.append(node)
  queue = [node]     # Initialize a queue inside the function

  while queue:    # Creating loop to visit each node
    m = queue.pop(0)
    print (m, end = " ")

    for neighbour in graph[m]:    # functions of BFS
      if neighbour not in visited:
        visited.append(neighbour)
        queue.append(neighbour)

# Driver Code
print("Following is the Breadth-First Search (BFS)")
visited = []   #List of visited nodes.
bfs(visited, graph, '5')    # Function calling
```

**DEPTH FIRST SEARCH (DFS)**

```
graph={
    '5' : ['3','7'],
    '3' : ['2', '4'],
    '7' : ['8'],
    '2' : [],
    '4' : ['8'],
    '8' : []
}
visited = set()  #set to keep track of visited nodes of graph
def dfs (visited, graph, node):  #function for dfs
  if node not in visited:
    print (node)
    visited.add(node)    # Add the node to the visited set
    for neighbour in graph[node]:
      dfs(visited, graph, neighbour)

# Driver code
print("Following is the Depth-First Search")
dfs(visited, graph, '5')  # Call the dfs function
```

```
graph={
    '5' : ['3','7'],
    '3' : ['2', '4'],
    '7' : ['8'],
    '2' : [],
    '4' : ['8'],
    '8' : []
}

visited = []
goal_node = '4'
def dfs (visited, graph, node):  #function for dfs
  if node not in visited:
    print (node)
    visited.append(node)         # Add the node to the visited set
    if node == goal_node:
      print(f'\ngoal node {goal_node} found')
      return True
    for neighbour in graph[node]:
      if dfs(visited, graph, neighbour):
        return True             # Propagate the True if goal is found in a recursive call
  return False                  # Return False if the node is visited or goal not found in this branch

#Driver code
print("Following is the Depth-First Search")
dfs(visited, graph, '5')       # Call the dfs function with a valid starting node
```

**Alpha-Beta Pruning**

```
tree = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F', 'G'],
    'D': [1, 2],
    'E': [3, 4],
    'F': [5, 6],
    'G': [7, 8]
}

def is_leaf(node):
    return isinstance(node, list) and all(isinstance(x, (int, float)) for x in node)

def alphabeta(node, depth, alpha, beta, maximizing):

    if is_leaf(tree[node]):
        return max(tree[node]) if maximizing else min(tree[node])

    if maximizing:
        value = float('-inf')
        for child in tree[node]:
            child_val = alphabeta(child, depth + 1, alpha, beta, False)
            value = max(value, child_val)
            alpha = max(alpha, value)
            if alpha >= beta:
                break
        return value
    else:
        value = float('inf')
        for child in tree[node]:
            child_val = alphabeta(child, depth + 1, alpha, beta, True)
            value = min(value, child_val)
            beta = min(beta, value)
            if beta <= alpha:
                break
        return value

result = alphabeta('A', 0, float('-inf'), float('inf'), True)
print("Optimal value:", result)
```

**Que : Write a program to impliment tower of hanoi.**

```
def tower_of_hanoi(n, source, auxiliary, destination):
    if n == 1:
      # Base case : Only 1 disk to move
        print(f"Move disk 1 from {source} to {destination}")
        return
    # Move n-1 disk from source to auxiliary , so they are out of the way
    tower_of_hanoi(n - 1, source, destination, auxiliary)
    # Move the nth disk from source to destination
    print(f"Move disk {n} from {source} to {destination}")
    # Move n-1 disk from auxiliary to destination
    tower_of_hanoi(n - 1, auxiliary, source, destination)

# Test the function with 3 disk
n = 3 # Number of disks
tower_of_hanoi(n, 'A', 'B', 'C')
```
